## 3. Öznitelik Bazında Veri Kalitesi Analizi

Bu bölümde her değişkenin saklama tipi, eksik değer oranı,
farklı değer sayısı ve sıfır oranı incelenmektedir.

Amaç, veri ön işleme aşamasından önce olası veri kalitesi
problemlerini belirlemektir.

```python
# Her öznitelik için veri kalitesi bilgileri
kalite_sonuclari = []

for sutun in df.columns:

    # Sütunun saklama tipi
    veri_tipi = str(df[sutun].dtype)

    # Gerçek NaN değerlerinin oranı
    eksik_orani = df[sutun].isna().mean() * 100

    # Farklı değer sayısı
    farkli_deger = df[sutun].nunique(dropna=True)

    # Sıfır oranı (yalnızca sayısal değişkenlerde)
    if pd.api.types.is_numeric_dtype(df[sutun]):
        sifir_orani = (df[sutun] == 0).mean() * 100
    else:
        sifir_orani = float("nan")

    kalite_sonuclari.append({
        "Öznitelik": sutun,
        "Saklama tipi": veri_tipi,
        "Eksik oranı (%)": eksik_orani,
        "Farklı değer sayısı": farkli_deger,
        "Sıfır oranı (%)": sifir_orani
    })

# Sonuçları tabloya dönüştür
kalite_tablosu = pd.DataFrame(kalite_sonuclari)

display(kalite_tablosu.round(2))
```

```python
# Kategorik sütunları belirle
kategorik_sutunlar = df.select_dtypes(
    include=["object"]
).columns

print("UNKNOWN DEĞERLERİ")
print("-" * 45)

for sutun in kategorik_sutunlar:

    adet = (df[sutun] == "unknown").sum()
    oran = adet / len(df) * 100

    if adet > 0:
        print(
            f"{sutun}: {adet} kayıt (%{oran:.2f})"
        )


print("\nPDAYS ÖZEL DEĞER KONTROLÜ")
print("-" * 45)

pdays_eksi_bir = (df["pdays"] == -1).sum()

print("pdays = -1:", pdays_eksi_bir)
print(
    "Oran:",
    round(pdays_eksi_bir / len(df) * 100, 2),
    "%"
)


print("\nPREVIOUS SIFIR KONTROLÜ")
print("-" * 45)

previous_sifir = (df["previous"] == 0).sum()

print("previous = 0:", previous_sifir)

# İki değişkenin ilişkisini incele
display(
    pd.crosstab(
        df["pdays"] == -1,
        df["previous"] == 0,
        rownames=["pdays = -1"],
        colnames=["previous = 0"]
    )
)
```

```python
# Önceki iletişim durumu ve kampanya sonucu
kontrol = pd.crosstab(
    df["pdays"] == -1,
    df["poutcome"] == "unknown",
    rownames=["pdays = -1"],
    colnames=["poutcome = unknown"]
)

display(kontrol)

# Önceden iletişim kurulmuş ancak
# kampanya sonucu bilinmeyen kayıtlar
supheli_kayitlar = df.loc[
    (df["previous"] > 0) &
    (df["poutcome"] == "unknown"),
    ["previous", "pdays", "poutcome"]
]

print("İncelenecek kayıt sayısı:", len(supheli_kayitlar))

display(supheli_kayitlar.head(10))
```

### Yorum – Veri Kalitesi ve Özel Değerler

Veri kümesinde doğrudan NaN olarak kaydedilmiş
eksik değer bulunmamaktadır. Ancak bazı kategorik
değişkenlerde "unknown" değerleri görülmektedir.

Özellikle contact (%28,80) ve poutcome (%81,75)
değişkenlerinde unknown oranları yüksektir.

previous değişkeninin %81,74'ü sıfırdır.
pdays değişkeninde ise aynı oranda -1 değeri
bulunmaktadır.

Çapraz tablo analizi sonucunda previous=0 ve

belirlenmiştir. Bu nedenle pdays=-1 değeri doğrudan

poutcome=unknown olan kayıtların beşinde önceki
iletişim sayısı sıfırdan büyüktür. Dolayısıyla
unknown değeri yalnızca önceki iletişimin
bulunmamasını temsil etmemektedir.

Bu bulgular, eksik değerlerin dönüştürülmesinden
önce değişkenlerin anlamlarının incelenmesi

Ölçüm ölçekleri, değişkenlerin anlamlarına ve veri

Unknown değerleri, değişkenlerin anlamları dikkate
alınarak değerlendirilecektir.

```python
# Orijinal veri kümesini koru
df_eksiklik = df.copy()

# Bilinmeyen kategorik değerler
eksik_sutunlar = ["job", "education", "contact"]

for sutun in eksik_sutunlar:
    df_eksiklik[sutun] = df_eksiklik[sutun].replace(
        "unknown", np.nan
    )

# Önceki iletişim gerçekleşmemişse
# önceki kampanya sonucu uygulanabilir değildir
iletisim_yok = df["previous"] == 0

df_eksiklik.loc[
    iletisim_yok, "poutcome"
] = "not_applicable"

# Önceki iletişim var fakat sonuç bilinmiyorsa
# eksik değer olarak işaretle
sonuc_bilinmiyor = (
    (df["previous"] > 0) &
    (df["poutcome"] == "unknown")
)

df_eksiklik.loc[
    sonuc_bilinmiyor, "poutcome"
] = np.nan

# Eksik değerlerin özetini oluştur
eksiklik_ozeti = pd.DataFrame({
    "Eksik sayısı": df_eksiklik.isna().sum(),
    "Eksik oranı (%)": (
        df_eksiklik.isna().mean() * 100
    )
})

eksiklik_ozeti = eksiklik_ozeti.sort_values(
    "Eksik oranı (%)", ascending=False
)

display(eksiklik_ozeti.round(2))

# Orijinal verinin değişmediğini kontrol et
assert df.isna().sum().sum() == 0

print("Orijinal veri korunmuştur.")
```

### 5.2 Eksiklik Desenlerinin İncelenmesi

incelenmektedir.

Amaç, eksiklikler arasında olası ilişkileri
belirlemektir.

### 5.3 Eksiklik Mekanizmasının İncelenmesi

Eksiklik mekanizmaları MCAR, MAR ve MNAR olarak
üç grupta incelenmektedir.

Bu aşamada contact değişkeninin eksiklik durumu
ile month değişkeni arasındaki ilişki araştırılacaktır.

Amaç, eksikliğin gözlemlenen değişkenlerden bağımsız
olup olmadığına ilişkin kanıt elde etmektir.

```python
from scipy.stats import chi2_contingency

# Contact sütununda eksik olan kayıtları belirle
contact_eksik = df_eksiklik["contact"].isna()

# Ay ve eksiklik durumu arasındaki çapraz tablo
ay_eksiklik = pd.crosstab(
    df["month"],
    contact_eksik
)

# Her iki durumu da tabloya dahil et
ay_eksiklik = ay_eksiklik.reindex(
    columns=[False, True],
    fill_value=0
)

# Aylara göre eksiklik oranlarını hesapla
ay_tablosu = pd.DataFrame({
    "Toplam kayıt": ay_eksiklik.sum(axis=1),
    "Eksik kayıt": ay_eksiklik[True],
    "Eksik oranı (%)": (
        ay_eksiklik[True]
        / ay_eksiklik.sum(axis=1) * 100
    )
})

# Eksiklik oranına göre sırala
ay_tablosu = ay_tablosu.sort_values(
    "Eksik oranı (%)",
    ascending=False
)

display(ay_tablosu.round(2))

# Ki-kare bağımsızlık testi
chi2, p_degeri, sd, beklenen = chi2_contingency(
    ay_eksiklik
)

# Etki büyüklüğü: Cramer's V
n = ay_eksiklik.to_numpy().sum()

cramers_v = np.sqrt(
    chi2 / (
        n * min(ay_eksiklik.shape[0] - 1,
                ay_eksiklik.shape[1] - 1)
    )
)

print("Ki-kare:", round(chi2, 4))
print("Serbestlik derecesi:", sd)
print("p-değeri:", p_degeri)
print("Cramer's V:", round(cramers_v, 4))
print("Minimum beklenen hücre sayısı:", round(beklenen.min(), 2))
```

### 5.4 Eğitim Bilgisindeki Eksikliklerin İncelenmesi

Bu aşamada education değişkeninin eksiklik durumu
ile job değişkeni arasındaki ilişki araştırılmaktadır.

Amaç, eğitim bilgisinin eksik olmasının farklı
meslek gruplarında aynı oranda görülüp
görülmediğini incelemektir.

```python
# Eğitim bilgisi eksik olan kayıtlar
education_eksik = df_eksiklik["education"].isna()

# Meslek ve eğitim eksikliği çapraz tablosu
meslek_eksiklik = pd.crosstab(
    df["job"],
    education_eksik
)

meslek_eksiklik = meslek_eksiklik.reindex(
    columns=[False, True],
    fill_value=0
)

# Mesleklere göre eksiklik oranlarını hesapla
meslek_tablosu = pd.DataFrame({
    "Toplam kayıt": meslek_eksiklik.sum(axis=1),
    "Eksik kayıt": meslek_eksiklik[True],
    "Eksik oranı (%)": (
        meslek_eksiklik[True]
        / meslek_eksiklik.sum(axis=1) * 100
    )
})

meslek_tablosu = meslek_tablosu.sort_values(
    "Eksik oranı (%)",
    ascending=False
)

display(meslek_tablosu.round(2))

# Ki-kare bağımsızlık testi
chi2, p_degeri, sd, beklenen = chi2_contingency(
    meslek_eksiklik
)

n = meslek_eksiklik.to_numpy().sum()

cramers_v = np.sqrt(
    chi2 / (
        n * min(
            meslek_eksiklik.shape[0] - 1,
            meslek_eksiklik.shape[1] - 1
        )
    )
)

print("Ki-kare:", round(chi2, 4))
print("Serbestlik derecesi:", sd)
print("p-değeri:", p_degeri)
print("Cramer's V:", round(cramers_v, 4))
print("Minimum beklenen:", round(beklenen.min(), 2))
```

## 6. Eksiklik Göstergesinin Model Başarımına Etkisi

### 6.1 Modelleme Verisinin Hazırlanması

Eksik değer göstergelerinin etkisini karşılaştırmak
için veri geliştirme ve test kümelerine ayrılacaktır.

Veri kümesinin orijinal satır sırası korunacak,
ilk %80 geliştirme, son %20 ise ileri dönem test
verisi olarak ayrılacaktır.

Duration değişkeni görüşme sonrasında bilindiği
için model girdilerinden çıkarılacaktır.

Test verisi model seçimi ve ön işleme
parametrelerinin öğrenilmesinde kullanılmayacaktır.

```python
# Modelleme verisini oluştur
X = df_eksiklik.drop(columns=["y", "duration"])

# Hedef değişkeni ikili hale getir
y = (df_eksiklik["y"] == "yes").astype(int)

# İlk %80 geliştirme, son %20 ileri dönem testi
ayrim = int(len(X) * 0.80)

X_gelistirme = X.iloc[:ayrim].copy()
X_test = X.iloc[ayrim:].copy()

y_gelistirme = y.iloc[:ayrim].copy()
y_test = y.iloc[ayrim:].copy()

# Veri boyutlarını göster
print("Geliştirme verisi:", X_gelistirme.shape)
print("Test verisi:", X_test.shape)

# Pozitif sınıf oranlarını incele
print(
    "Geliştirme pozitif oranı (%):",
    round(y_gelistirme.mean() * 100, 2)
)

print(
    "Test pozitif oranı (%):",
    round(y_test.mean() * 100, 2)
)

# Kontroller
assert len(X_gelistirme) + len(X_test) == len(df)

assert "duration" not in X_gelistirme.columns
assert "y" not in X_gelistirme.columns

print("Veri ayrımı doğrulandı.")
```

### 6.2 Eksik Veri Doldurma ve Kodlama Hatları

Sayısal değişkenlerde eksik değerler medyan,
kategorik değişkenlerde ise en sık görülen
kategori kullanılarak doldurulacaktır.

Kategorik değişkenler One-Hot Encoding ile
sayısal gösterime dönüştürülecektir.

Eksiklik göstergelerinin etkisini incelemek için
add_indicator=False ve add_indicator=True
seçenekleri karşılaştırılacaktır.

Ön işleme adımları model değerlendirmesinde
Pipeline içerisinde kullanılacaktır.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Sayısal ve kategorik değişkenleri belirle
sayisal_sutunlar = X_gelistirme.select_dtypes(
    include=np.number
).columns.tolist()

kategorik_sutunlar = X_gelistirme.select_dtypes(
    exclude=np.number
).columns.tolist()

print("Sayısal sütunlar:", sayisal_sutunlar)
print("Kategorik sütunlar:", kategorik_sutunlar)


# Ön işleme hattını oluşturan fonksiyon
def on_isleme_olustur(gosterge=False):

    # Sayısal değişkenler
    sayisal_hat = Pipeline([
        (
            "doldur",
            SimpleImputer(
                strategy="median",
                add_indicator=gosterge
            )
        ),
        ("olcekle", StandardScaler())
    ])

    # Kategorik değişkenler
    kategorik_hat = Pipeline([
        (
            "doldur",
            SimpleImputer(
                strategy="most_frequent",
                add_indicator=gosterge
            )
        ),
        (
            "kodla",
            OneHotEncoder(handle_unknown="ignore")
        )
    ])

    return ColumnTransformer([
        ("sayisal", sayisal_hat, sayisal_sutunlar),
        ("kategorik", kategorik_hat, kategorik_sutunlar)
    ])


# İki farklı ön işleme hattı
on_isleme_normal = on_isleme_olustur(False)
on_isleme_gostergeli = on_isleme_olustur(True)

print("\nİki ön işleme hattı oluşturuldu.")
print("Normal:", on_isleme_normal is not None)
print("Göstergeli:", on_isleme_gostergeli is not None)
```

### 6.3 Eksiklik Göstergesi Karşılaştırması

Eksiklik göstergelerinin model performansına etkisi,
aynı lojistik regresyon modeli kullanılarak
incelenmektedir.

İki model aynı geliştirme verisinde, aynı çapraz
doğrulama bölmeleriyle karşılaştırılacaktır.

Değerlendirmede 5 katlı, 2 tekrarlı katmanlı
çapraz doğrulama ve Average Precision kullanılacaktır.

İleri dönem test verisi bu deneyde kullanılmayacaktır.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import (
    RepeatedStratifiedKFold,
    cross_val_score
)

# Her iki model için aynı değerlendirme protokolü
cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=2,
    random_state=SEED
)

# Karşılaştırılacak ön işleme hatları
deneyler = {
    "Göstergesiz": on_isleme_normal,
    "Göstergeli": on_isleme_gostergeli
}

sonuclar = {}

for deney_adi, on_isleme in deneyler.items():

    # Ön işleme ve model aynı Pipeline içinde
    model = Pipeline([
        ("on_isleme", on_isleme),
        ("siniflandirici", LogisticRegression(
            max_iter=2000,
            random_state=SEED
        ))
    ])

    # Çapraz doğrulama
    skorlar = cross_val_score(
        model,
        X_gelistirme,
        y_gelistirme,
        cv=cv,
        scoring="average_precision",
        n_jobs=1
    )

    sonuclar[deney_adi] = skorlar

    print(
        f"{deney_adi}: "
        f"AP = {skorlar.mean():.4f} "
        f"± {skorlar.std():.4f}"
    )

# Aynı katlardaki performans farkları
farklar = (
    sonuclar["Göstergeli"]
    - sonuclar["Göstergesiz"]
)

print("\nGöstergenin AP farkı:")
print(f"Ortalama fark: {farklar.mean():+.4f}")
print(f"Farkların std değeri: {farklar.std():.4f}")
```

### 7.1 Kodlama İçin Öznitelik Seçimi

One-Hot Encoding, nadir kategori birleştirme ve

uygun bir öznitelik belirlenecektir.

### 7.2 Standart One-Hot Encoding

Kodlama karşılaştırması için day değişkeni
seçilmiştir.

Day değişkeni takvimsel bir özellik taşımasına
rağmen bu deney kapsamında kategorik olarak
ele alınacaktır.

Diğer değişkenlerin ön işleme yöntemleri
sabit tutulacaktır.

One-Hot Encoding işlemi Pipeline içerisinde
ve yalnızca eğitim katlarında öğrenilecektir.

```python
# Day değişkenini sayısal sütunlardan çıkar
sayisal_day_disinda = [
    sutun for sutun in sayisal_sutunlar
    if sutun != "day"
]

# Her özniteliğin yalnızca bir grupta olduğunu kontrol et
tum_sutunlar = (
    sayisal_day_disinda
    + kategorik_sutunlar
    + ["day"]
)

assert len(tum_sutunlar) == len(set(tum_sutunlar))
assert set(tum_sutunlar) == set(X_gelistirme.columns)

print("Sayısal sütunlar:", sayisal_day_disinda)
print("Kategorik sütunlar:", kategorik_sutunlar)
print("Özel kodlanacak sütun: day")
print("Öznitelik grupları doğrulandı.")

# Gün değişkeninin kodlayıcısı değiştirilebilir.
# Diğer ön işleme adımları sabit kalır.

def gun_kodlama_hatti(kodlayici):

    sayisal_hat = Pipeline([
        ("doldur", SimpleImputer(
            strategy="median",
            add_indicator=True
        )),
        ("olcekle", StandardScaler())
    ])

    kategorik_hat = Pipeline([
        ("doldur", SimpleImputer(
            strategy="most_frequent",
            add_indicator=True
        )),
        ("kodla", OneHotEncoder(
            handle_unknown="ignore"
        ))
    ])

    gun_hat = Pipeline([
        ("doldur", SimpleImputer(
            strategy="most_frequent"
        )),
        ("kodla", kodlayici)
    ])

    return ColumnTransformer([
        ("sayisal", sayisal_hat, sayisal_day_disinda),
        ("kategorik", kategorik_hat, kategorik_sutunlar),
        ("gun", gun_hat, ["day"])
    ])
```

```python
# Standart One-Hot Encoding
on_isleme_ohe = gun_kodlama_hatti(
    OneHotEncoder(handle_unknown="ignore")
)

# Lojistik regresyon modeli
model_ohe = Pipeline([
    ("on_isleme", on_isleme_ohe),
    ("siniflandirici", LogisticRegression(
        max_iter=2000,
        random_state=SEED
    ))
])

# Tekrarlı çapraz doğrulama
skor_ohe = cross_val_score(
    model_ohe,
    X_gelistirme,
    y_gelistirme,
    cv=cv,
    scoring="average_precision",
    n_jobs=1
)

print("Standart One-Hot Encoding")
print(f"AP ortalama: {skor_ohe.mean():.4f}")
print(f"AP std: {skor_ohe.std():.4f}")
print("CV değerlendirme sayısı:", len(skor_ohe))
```

### 7.3 Nadir Kategorileri Birleştiren One-Hot Encoding

Standart One-Hot Encoding yönteminde her kategori
ayrı bir sütunla temsil edilmektedir.

Bu deneyde eğitim verisindeki görülme sıklığı
%2'nin altında olan kategoriler birleştirilecektir.

Amaç, nadir kategorilerin ayrı ayrı temsil edilmesi
ile birleştirilmesinin model başarımına etkisini
karşılaştırmaktır.

Diğer ön işleme adımları ve değerlendirme
protokolü değiştirilmemiştir.

```python
# Nadir kategorileri birleştiren One-Hot Encoding
on_isleme_nadir = gun_kodlama_hatti(
    OneHotEncoder(
        min_frequency=0.02,
        handle_unknown="infrequent_if_exist"
    )
)

# Aynı sınıflandırıcı
model_nadir = Pipeline([
    ("on_isleme", on_isleme_nadir),
    ("siniflandirici", LogisticRegression(
        max_iter=2000,
        random_state=SEED
    ))
])

# Aynı CV protokolü
skor_nadir = cross_val_score(
    model_nadir,
    X_gelistirme,
    y_gelistirme,
    cv=cv,
    scoring="average_precision",
    n_jobs=1
)

print("Nadir Kategori Birleştirmeli One-Hot")
print(f"AP ortalama: {skor_nadir.mean():.4f}")
print(f"AP std: {skor_nadir.std():.4f}")
print("CV değerlendirme sayısı:", len(skor_nadir))

# Standart One-Hot ile karşılaştırma
fark = skor_nadir - skor_ohe

print("\nStandart One-Hot ile karşılaştırma")
print(f"Standart OHE: {skor_ohe.mean():.4f}")
print(f"Nadir OHE: {skor_nadir.mean():.4f}")
print(f"Ortalama AP farkı: {fark.mean():+.4f}")
print(f"Farkların std değeri: {fark.std():.4f}")
```

### 7.4 Kat Dışı Target Encoding

Target Encoding yönteminde kategoriler, hedef
değişkene ilişkin istatistiklerle sayısal olarak
temsil edilmektedir.

Bu yaklaşım, kategoriler ve hedef değişken
arasındaki ilişkilerin öğrenilmesini sağlar.

Ancak hedef bilgilerinin doğrudan aynı kayıtların
kodlanmasında kullanılması veri sızıntısına
neden olabilir.

Bu nedenle TargetEncoder, Pipeline içerisinde
ve iç katlamalı cross-fitting yöntemiyle
kullanılacaktır.

Karşılaştırmada önceki deneylerle aynı
değerlendirme protokolü korunacaktır.

```python
from sklearn.preprocessing import TargetEncoder

# Kat dışı hedef kodlama
on_isleme_hedef = gun_kodlama_hatti(
    TargetEncoder(
        target_type="binary",
        smooth="auto",
        cv=5,
        shuffle=True,
        random_state=SEED
    )
)

# Aynı lojistik regresyon modeli
model_hedef = Pipeline([
    ("on_isleme", on_isleme_hedef),
    ("siniflandirici", LogisticRegression(
        max_iter=2000,
        random_state=SEED
    ))
])

# Aynı çapraz doğrulama protokolü
skor_hedef = cross_val_score(
    model_hedef,
    X_gelistirme,
    y_gelistirme,
    cv=cv,
    scoring="average_precision",
    n_jobs=1
)

print("Kat Dışı Target Encoding")
print(f"AP ortalama: {skor_hedef.mean():.4f}")
print(f"AP std: {skor_hedef.std():.4f}")
print("CV değerlendirme sayısı:", len(skor_hedef))
```

### 7.5 Hatalı Target Encoding ve Veri Sızıntısı

Bu deneyde Target Encoding işlemi kasıtlı olarak
çapraz doğrulamadan önce gerçekleştirilecektir.

Kodlayıcı bütün geliştirme verisinin hedef
değerlerini kullanarak öğrenilecektir.

Böylece çapraz doğrulamanın doğrulama katlarındaki
hedef bilgileri de kodlama aşamasında kullanılmış
olacaktır.

Bu kurulum metodolojik olarak hatalıdır ve
yalnızca veri sızıntısını göstermek amacıyla
uygulanmaktadır.

Elde edilen skor geçerli model performansı
olarak değerlendirilmeyecektir.

```python
# HATALI YÖNTEM: Target Encoding CV öncesinde öğreniliyor

yanlis_encoder = TargetEncoder(
    target_type="binary",
    smooth="auto",
    random_state=SEED
)

# Bütün geliştirme verisinin hedeflerini görüyor
yanlis_encoder.fit(
    X_gelistirme[["day"]],
    y_gelistirme
)

# Aynı veriyi, kendi hedef bilgilerini de içeren
# istatistiklerle kodluyor
X_yanlis = X_gelistirme.copy()

X_yanlis["day"] = yanlis_encoder.transform(
    X_gelistirme[["day"]]
)[:, 0]

print("Hatalı kodlama tamamlandı.")
print("Kodlanmış ilk 5 day değeri:")

display(X_yanlis[["day"]].head())

# Diğer değişkenlerin ön işlemesi Pipeline içinde kalır

sayisal_hat_yanlis = Pipeline([
    ("doldur", SimpleImputer(
        strategy="median",
        add_indicator=True
    )),
    ("olcekle", StandardScaler())
])

kategorik_hat_yanlis = Pipeline([
    ("doldur", SimpleImputer(
        strategy="most_frequent",
        add_indicator=True
    )),
    ("kodla", OneHotEncoder(
        handle_unknown="ignore"
    ))
])

on_isleme_yanlis = ColumnTransformer([
    (
        "sayisal",
        sayisal_hat_yanlis,
        sayisal_day_disinda + ["day"]
    ),
    (
        "kategorik",
        kategorik_hat_yanlis,
        kategorik_sutunlar
    )
])

model_yanlis = Pipeline([
    ("on_isleme", on_isleme_yanlis),
    ("siniflandirici", LogisticRegression(
        max_iter=2000,
        random_state=SEED
    ))
])

# Aynı CV protokolü; fakat day zaten sızıntılı
skor_yanlis = cross_val_score(
    model_yanlis,
    X_yanlis,
    y_gelistirme,
    cv=cv,
    scoring="average_precision",
    n_jobs=1
)

print("\nDOĞRU VE HATALI TARGET ENCODING")
print(
    f"Doğru:  {skor_hedef.mean():.4f}"
    f" ± {skor_hedef.std():.4f}"
)
print(
    f"Hatalı: {skor_yanlis.mean():.4f}"
    f" ± {skor_yanlis.std():.4f}"
)

fark = skor_yanlis - skor_hedef

print(f"Ortalama fark: {fark.mean():+.4f}")
print(f"Farkların std değeri: {fark.std():.4f}")
```

## 8. Ön İşleme Kaynaklı Veri Sızıntısı

### 8.1 Pipeline Dışında Ön İşleme

Bu deneyde eksik değer doldurma, ölçekleme ve
kategorik kodlama işlemleri çapraz doğrulamadan
önce bütün geliştirme verisine uygulanacaktır.

Bu yaklaşım metodolojik olarak hatalıdır.
Çünkü doğrulama katlarının öznitelik bilgileri
ön işleme parametrelerinin hesaplanmasına
dahil edilmektedir.

Hatalı kurulum yalnızca doğru Pipeline
yaklaşımıyla karşılaştırma amacıyla kullanılacaktır.

İleri dönem test verisi kullanılmayacaktır.

```python
from sklearn.base import clone

# HATALI KURULUM
# Ön işleme bütün geliştirme verisinde öğreniliyor.

yanlis_on_isleme = clone(on_isleme_gostergeli)

X_onceden_islenmis = yanlis_on_isleme.fit_transform(
    X_gelistirme
)

print("Ön işleme CV dışında tamamlandı.")
print("İşlenmiş veri boyutu:", X_onceden_islenmis.shape)

# Ön işleme artık Pipeline içinde değil.
# Sadece lojistik regresyon CV ile değerlendiriliyor.

model_yanlis_on_isleme = LogisticRegression(
    max_iter=2000,
    random_state=SEED
)

skor_yanlis_on_isleme = cross_val_score(
    model_yanlis_on_isleme,
    X_onceden_islenmis,
    y_gelistirme,
    cv=cv,
    scoring="average_precision",
    n_jobs=1
)

# Önceki doğru Pipeline deneyinin skorları
skor_dogru_on_isleme = sonuclar["Göstergeli"]

# İki deneyin sonuçları
print("\nDOĞRU VE HATALI ÖN İŞLEME")

print(
    f"Doğru Pipeline: "
    f"{skor_dogru_on_isleme.mean():.4f} "
    f"± {skor_dogru_on_isleme.std():.4f}"
)

print(
    f"Hatalı ön işleme: "
    f"{skor_yanlis_on_isleme.mean():.4f} "
    f"± {skor_yanlis_on_isleme.std():.4f}"
)

# Aynı katlar üzerindeki farklar
farklar_on_isleme = (
    skor_yanlis_on_isleme - skor_dogru_on_isleme
)

print(
    f"Ortalama AP farkı: "
    f"{farklar_on_isleme.mean():+.4f}"
)

print(
    f"Farkların std değeri: "
    f"{farklar_on_isleme.std():.4f}"
)
```

### 10.4 Weka İçin ARFF Veri Dosyası

Weka deneylerinde Python ile aynı geliştirme
kümesi kullanılacaktır.

Girdi değişkenleri ve hedef sütunu ARFF
biçimine dönüştürülecektir.

Duration değişkeni görüşme sonrasında bilindiği
için dosyaya dahil edilmeyecektir.

Eksik değerler ARFF standardına uygun olarak
? ile gösterilecektir.

Bu aşamada eksik değer doldurma veya
ölçekleme yapılmayacaktır.

```python
# Weka için geliştirme verisini hazırla
veri_weka = X_gelistirme.copy()

# Hedef değişkeni son sütuna ekle
veri_weka["y"] = y_gelistirme.map({
    0: "no",
    1: "yes"
})

# Sayısal ve kategorik değişkenleri belirle
sayisal_weka = veri_weka.select_dtypes(
    include=np.number
).columns.tolist()

kategorik_weka = [
    sutun for sutun in veri_weka.columns
    if sutun not in sayisal_weka
]

# Temel kontroller
assert len(veri_weka) == 36168
assert veri_weka.shape[1] == 16
assert veri_weka.columns[-1] == "y"
assert "duration" not in veri_weka.columns
assert veri_weka["y"].notna().all()

print("Weka veri boyutu:", veri_weka.shape)
print("Sayısal sütunlar:", sayisal_weka)
print("Kategorik sütunlar:", kategorik_weka)

# ARFF kategorik değerlerini güvenli biçimde yaz
def arff_kategori(deger):
    metin = str(deger)
    metin = metin.replace("\\", "\\\\")
    metin = metin.replace("'", "\\'")
    return f"'{metin}'"


arff_yolu = weka_klasoru / "bank_gelistirme.arff"

with open(arff_yolu, "w", encoding="utf-8") as f:

    f.write("@RELATION bank_marketing_development\n\n")

    # Değişken tanımları
    for sutun in veri_weka.columns:

        if sutun in sayisal_weka:
            f.write(f"@ATTRIBUTE '{sutun}' NUMERIC\n")

        else:
            if sutun == "y":
                kategoriler = ["no", "yes"]
            else:
                kategoriler = sorted(
                    veri_weka[sutun]
                    .dropna()
                    .astype(str)
                    .unique()
                )

            degerler = ",".join(
                arff_kategori(k)
                for k in kategoriler
            )

            f.write(
                f"@ATTRIBUTE '{sutun}' {{{degerler}}}\n"
            )

    f.write("\n@DATA\n")

    # Kayıtları yaz
    for satir in veri_weka.itertuples(
        index=False, name=None
    ):

        degerler = []

        for sutun, deger in zip(
            veri_weka.columns, satir
        ):

            if pd.isna(deger):
                degerler.append("?")

            elif sutun in sayisal_weka:
                degerler.append(str(deger))

            else:
                degerler.append(
                    arff_kategori(deger)
                )

        f.write(",".join(degerler) + "\n")


print("\nARFF dosyası oluşturuldu.")
print("Dosya:", arff_yolu)
print(
    "Boyut (MB):",
    round(arff_yolu.stat().st_size / 1024**2, 2)
)
```

### 10.8 Weka Sonuçlarının Karşılaştırılması

ZeroR ve J48 algoritmaları aynı geliştirme
verisi üzerinde 10 katlı çapraz doğrulama
ile değerlendirilmiştir.

Sınıf dengesizliği nedeniyle yalnızca accuracy
değil, pozitif sınıfa ait precision, recall,
ROC Area ve PRC Area sonuçları incelenmektedir.

Weka sonuçları, Python'daki baseline ve ilk
model deneylerinin bağımsız kontrolü amacıyla
raporlanmaktadır.

```python
import re

def weka_sonuc_ozeti(cikti, model_adi):

        # Yalnızca çapraz doğrulama sonuçlarını kullan
    cv_baslik = "=== Stratified cross-validation ==="

    if cv_baslik not in cikti:
        raise ValueError(
            "Weka çapraz doğrulama bölümü bulunamadı."
        )

    cikti = cikti.split(cv_baslik, 1)[1]

    # Accuracy değerini bul
    eslesme = re.search(
        r"Correctly Classified Instances\s+\d+\s+([\d.]+)\s*%",
        cikti
    )

    if not eslesme:
        raise ValueError("Accuracy bulunamadı.")

    accuracy = float(eslesme.group(1))

    # Yalnızca sınıf bazlı metrikler bölümünü al
    baslik = "=== Detailed Accuracy By Class ==="
    bitis = "=== Confusion Matrix ==="

    if baslik not in cikti or bitis not in cikti:
        raise ValueError("Weka metrik tablosu bulunamadı.")

    metrik_bolumu = (
        cikti.split(baslik, 1)[1]
        .split(bitis, 1)[0]
    )

    # Yalnızca yes sınıfının metrik satırını seç
    yes_satiri = None

    for satir in metrik_bolumu.splitlines():

        alanlar = satir.strip().split()

        if len(alanlar) == 9 and alanlar[-1] == "yes":
            yes_satiri = alanlar
            break

    if yes_satiri is None:
        raise ValueError("yes sınıfının metrikleri bulunamadı.")

    def sayiya_cevir(deger):
        if deger == "?":
            return float("nan")
        return float(deger)

    return {
        "Model": model_adi,
        "Accuracy (%)": accuracy,
        "yes Precision": sayiya_cevir(yes_satiri[2]),
        "yes Recall": sayiya_cevir(yes_satiri[3]),
        "yes F1": sayiya_cevir(yes_satiri[4]),
        "ROC AUC": sayiya_cevir(yes_satiri[6]),
        "yes PRC Area": sayiya_cevir(yes_satiri[7])
    }


weka_karsilastirma = pd.DataFrame([
    weka_sonuc_ozeti(zeror_sonuc.stdout, "ZeroR"),
    weka_sonuc_ozeti(j48_sonuc.stdout, "J48")
])

display(weka_karsilastirma.round(4))

print("Weka karşılaştırma tablosu oluşturuldu.")
```

Bu nedenle söz konusu değerler `NaN` biçimine
dönüştürülmüştür.

Bu dönüşümün amacı, eksik bilgileri gerçek
kategorilerden ayırmak ve eksiklik
göstergelerinin etkisini inceleyebilmektir.

**Karar 5 – `pdays` ve `poutcome` özel durumları**

`pdays=-1` değeri, müşteriyle önceki kampanyada
iletişim kurulmadığını gösteren özel bir koddur.

İnceleme sonucunda bu değerin `previous=0`
durumuyla örtüştüğü görülmüştür.

Dolayısıyla `pdays=-1` doğrudan eksik değer
olarak değerlendirilmemiş, korunmuştur.

Benzer biçimde, önceki iletişimi bulunmayan
müşterilerin `poutcome=unknown` değerleri
`not_applicable` olarak yeniden kodlanmıştır.

Önceki iletişimi bulunduğu hâlde sonucu
bilinmeyen beş kaydın `poutcome` değeri ise
gerçek eksik bilgi olarak ele alınmıştır.

Bu ayrım, uygulanamaz bir durumun eksik
veriyle karıştırılmasını önlemek amacıyla
yapılmıştır.

**Karar 6 – Eksiklik mekanizmalarının değerlendirilmesi**

Eksikliklerin tamamen rastgele oluşup
oluşmadığı, eksiklik göstergeleri ile
gözlenen değişkenler arasındaki ilişkiler
üzerinden incelenmiştir.

`contact` eksikliği ile `month` arasında
güçlü bir ilişki bulunmuştur
(Cramér's V = 0,7147).

`education` eksikliği ile `job` eksikliği
arasında da ilişki gözlenmiştir
(Cramér's V = 0,1948).

Bu bulgular, incelenen değişkenlerde MCAR
(tamamen rastgele eksiklik) varsayımını
zayıflatmaktadır.

Eksikliklerin gözlenen değişkenlere bağlı
olması MAR mekanizmasıyla uyumludur.

Ancak yalnızca gözlenen verilerle MAR
mekanizması kesin olarak kanıtlanamaz;
MNAR olasılığı da dışlanamaz.

Bu nedenle eksiklik mekanizması hakkında
kesin hüküm verilmemiş, sonuçlar kanıta
dayalı ve sınırlılıkları belirtilerek
yorumlanmıştır.



### 11.3 Eksiklik Göstergeleri ve Kategorik Kodlama

**Karar 7 – Eksiklik göstergelerinin kullanılması**

Eksik değerlerin yalnızca doldurulması yerine,
eksikliğin kendisinin ek bilgi taşıyıp
taşımadığı da incelenmiştir.

Bu amaçla `SimpleImputer` sınıfının
`add_indicator` parametresi karşılaştırılmıştır.

Eksiklik göstergeleri kullanılmadığında
ortalama AP `0,2155`, kullanıldığında ise
`0,2197` olarak ölçülmüştür.

Aynı çapraz doğrulama bölmeleri üzerinde
ortalama AP farkı yaklaşık `+0,0042` olmuştur.

Bu sonuç, incelenen veri ve model için
eksiklik göstergelerinin ek tahmin bilgisi
sağlayabildiğini göstermektedir.

Ancak bu fark, bütün modellerde aynı
iyileşmenin elde edileceği anlamına gelmez.

**Karar 8 – Kategorik kodlama yöntemlerinin karşılaştırılması**

Kategorik kodlamanın model performansına
etkisini incelemek için day değişkeni
kategorik olarak değerlendirilmiştir.

Bu değişkenin 31 farklı değeri bulunmaktadır.

Standart One-Hot Encoding, nadir kategori
gruplamalı One-Hot Encoding ve Target Encoding
yöntemleri karşılaştırılmıştır.

Ortalama AP sonuçları sırasıyla:

- Standart One-Hot Encoding: `0,2265`
- Nadir kategori gruplamalı OHE: `0,2267`
- Target Encoding: `0,2211`

olarak bulunmuştur.

Nadir kategori gruplamalı OHE ile standart
OHE arasındaki ortalama fark çok küçüktür.

Bu nedenle sonuçlar, nadir kategorileri
gruplamanın belirgin bir üstünlüğünü
göstermemektedir.

Day değişkeni aslında ayın gününü ifade
eden sayısal bir değişkendir. Bu deneyde
kodlama davranışını karşılaştırmak amacıyla
kategorik biçimde ele alınmıştır.

31 farklı değere sahip olması nedeniyle
çok yüksek kardinaliteli bir değişken
olarak değerlendirilmemelidir.

Bu tercih, deneyin önemli bir sınırlılığıdır.


**Ek doğrulama – Nadir kategori gruplaması**

Kodlama deneyinde kullanılan `day` değişkeni,
incelenen adaylar arasında 31 farklı değerle
en yüksek kardinaliteye sahiptir.

Geliştirme verisi üzerinde gerçekleştirilen
bağımsız kodlama kontrolünde, %2 frekans
eşiğinin altında altı kategori belirlenmiştir:
1, 10, 22, 24, 25 ve 31.

Standart One-Hot Encoding 31 çıktı sütunu
üretirken, nadir kategori gruplamalı
One-Hot Encoding 26 çıktı sütunu üretmiştir.

Böylece nadir kategorilerin tek grupta
birleştirildiği doğrulanmıştır.

Bu kontrol yalnızca kodlama davranışının
incelenmesi amacıyla yapılmıştır.
Model başarımı için çapraz doğrulama
içerisindeki Pipeline sonuçları esas alınmıştır.

Standart OHE ile nadir kategorileri birleştiren
OHE arasında ortalama AP farkı yaklaşık
0,0002 olduğundan, belirgin bir performans
üstünlüğü sonucuna ulaşılmamıştır.

`day` değişkeni mevcut veri kümesindeki
en yüksek kardinaliteli aday olmakla birlikte,
31 kategorinin genel anlamda çok yüksek
kardinaliteyi temsil etmediği çalışmanın
sınırlılığı olarak korunmuştur.

**Karar 9 – Target Encoding ve hedef sızıntısı**

Target Encoding, kategori değerlerini
hedef değişkenle ilişkili istatistikler
üzerinden sayısal olarak temsil etmektedir.

Bu nedenle kodlama işleminin çapraz
doğrulama sınırları içinde yapılması
gerekmektedir.

Doğru deneyde Target Encoding, Pipeline
içerisinde kullanılmış ve eğitim
bölmelerindeki iç çapraz kodlama
mekanizmasından yararlanılmıştır.

Böylece değerlendirme bölmesinin hedef
değerlerinin kodlama işlemine dahil
edilmesi önlenmiştir.

Karşılaştırma amacıyla kasıtlı olarak
hatalı bir deney de gerçekleştirilmiştir.

Hatalı yöntemde hedef kodlaması,
geliştirme verisinin tamamında
çapraz doğrulamadan önce uygulanmıştır.

Doğru yöntemde ortalama AP `0,2211`,
hatalı yöntemde ise `0,2247` olmuştur.

Hatalı yöntemin daha yüksek skor üretmesi,
bu yöntemin geçerli olduğu anlamına gelmez.

Çünkü doğrulama bölmelerine ait hedef
bilgisi, kodlama istatistiklerinin
hesaplanmasına dolaylı olarak katılmıştır.

Bu nedenle hatalı deneyin performans
sonucu model seçimi için kullanılmamıştır.

Bu karşılaştırma, veri sızıntısının
yalnızca model eğitimi sırasında değil,
özellik dönüşümü aşamasında da
oluşabileceğini göstermektedir.



### 11.4 Pipeline, Çapraz Doğrulama ve Baseline

**Karar 10 – Ön işlemenin Pipeline içerisinde gerçekleştirilmesi**

Eksik değer doldurma, kategorik kodlama ve
ölçekleme işlemlerinin çapraz doğrulama
sürecindeki konumu değerlendirilmiştir.

Doğru yaklaşımda ön işleme ve sınıflandırma
adımları scikit-learn Pipeline yapısında
birleştirilmiştir.

Böylece her çapraz doğrulama katında ön işleme
istatistikleri yalnızca ilgili eğitim
bölmesinden öğrenilmiştir.

Doğrulama bölmesi, eğitim bölmesinde öğrenilen
dönüşümler kullanılarak işlenmiştir.

Karşılaştırma amacıyla kasıtlı olarak
hatalı bir deney de gerçekleştirilmiştir.

Hatalı yaklaşımda ön işleme, geliştirme
verisinin tamamına çapraz doğrulamadan
önce uygulanmıştır.

Ortalama AP sonuçları:

- Hatalı ön işleme: `0,2196 ± 0,0101`
- Pipeline içinde doğru ön işleme:
  `0,2197 ± 0,0099`

Aynı bölmeler üzerinden hesaplanan
ortalama fark yaklaşık `-0,0001` olmuştur.

Hatalı yöntemin daha yüksek skor üretmemesi,
veri sızıntısı riskinin bulunmadığı
anlamına gelmez.

Çünkü doğrulama bölmesindeki özellik
dağılımları, ön işleme istatistiklerinin
hesaplanmasını etkilemiştir.

Bu nedenle Pipeline yaklaşımı tercih
edilmiş, hatalı deney yalnızca
yöntemsel karşılaştırma amacıyla
kullanılmıştır.

**Karar 11 – Çapraz doğrulama protokolü**

Geliştirme kümesindeki Python deneylerinde
RepeatedStratifiedKFold kullanılmıştır.

Çapraz doğrulama 5 kat ve 2 tekrar
olacak şekilde uygulanmıştır.

Rastgelelik tohumu `42` olarak belirlenmiştir.

Katmanlı örnekleme, her katın sınıf
dağılımının geliştirme kümesindeki
dağılıma yakın kalmasını amaçlamaktadır.

Tekrarlı değerlendirme, farklı veri
bölmelerindeki performans değişkenliğinin
incelenmesine olanak sağlamıştır.

Sonuçlar ortalama AP ve katlar arasındaki
standart sapma ile raporlanmıştır.

Ancak aynı örnekler tekrarlar boyunca
değerlendirildiğinden kat skorları
tamamen bağımsız gözlemler değildir.

Ayrıca rastgele katmanlı çapraz doğrulama,
verinin kronolojik sırasını korumamaktadır.

Dolayısıyla bu sonuçlar ileri dönemdeki
başarımın doğrudan ölçümü olarak
yorumlanmamıştır.

**Karar 12 – DummyClassifier ile baseline oluşturulması**

Bir modelin başarımını değerlendirebilmek
için basit bir referans modele ihtiyaç
duyulmuştur.

Bu amaçla DummyClassifier
(strategy="prior") kullanılmıştır.

Model, özniteliklerden tahmin ilişkisi
öğrenmek yerine eğitim bölmesinin
sınıf dağılımını temel almaktadır.

DummyClassifier için ortalama AP
`0,067269 ± 0,000063` olarak ölçülmüştür.

Bu değer geliştirme verisindeki pozitif
sınıf oranıyla yaklaşık olarak aynıdır.

Dolayısıyla baseline, veri kümesindeki
sınıf dengesizliğini değerlendirmek
için bir referans sağlamıştır.

**Karar 13 – İlk model olarak lojistik regresyon**

İlk öğrenen model olarak LogisticRegression
tercih edilmiştir.

Lojistik regresyon, sınıflandırma
problemlerinde anlaşılabilir ve
tekrarlanabilir bir başlangıç
modeli sunmaktadır.

Model, ön işleme adımlarıyla birlikte
Pipeline içerisinde değerlendirilmiştir.

Ortalama AP değeri
`0,219723 ± 0,009936` olarak bulunmuştur.

DummyClassifier ile aynı çapraz doğrulama
bölmelerinde karşılaştırıldığında
ortalama AP farkı yaklaşık `+0,152453`
olmuştur.

Bu bulgu, lojistik regresyonun geliştirme
kümesinde baseline'a göre daha iyi bir
pozitif sınıf sıralaması oluşturduğunu
göstermektedir.

Ancak bu sonuç, modelin ileri dönem test
verisinde aynı başarımı sağlayacağını
garanti etmemektedir.

Bölüm 1 kapsamında amaç kapsamlı
hiperparametre optimizasyonu değil,
sızıntısız değerlendirme sürecinin
kurulması ve ilk modelin baseline
ile karşılaştırılmasıdır.



### 11.5 Weka Kontrolü ve Deneylerin Sınırlılıkları

**Karar 14 – Weka ile bağımsız kontrol yapılması**

Python ortamında gerçekleştirilen değerlendirmelere
ek olarak Weka yazılımıyla bağımsız bir kontrol
yapılmıştır.

Bu amaçla geliştirme kümesi ARFF biçimine
dönüştürülmüş ve Weka ZeroR ile J48
sınıflandırıcıları çalıştırılmıştır.

Her iki Weka deneyinde `36.168` kayıtlık
geliştirme kümesi, 10 katlı katmanlı çapraz
doğrulama ve `42` rastgelelik tohumu kullanılmıştır.

Weka ZeroR için accuracy `%93,2731`,
pozitif sınıfın recall değeri `0,000`
ve PRC Area değeri `0,067` bulunmuştur.

ZeroR bütün kayıtları çoğunluk sınıfı olan
`"no"` olarak tahmin etmiştir.

Weka J48 için accuracy `%93,2537`,
pozitif sınıf precision değeri `0,494`,
recall değeri `0,113` ve PRC Area değeri
`0,188` olarak ölçülmüştür.

J48, `2.433` gerçek pozitif kaydın
275'ini doğru tahmin edebilmiştir.

Bu sonuçlar, yüksek accuracy değerinin
pozitif sınıfı başarılı biçimde tespit etmek
anlamına gelmediğini göstermektedir.

**Karar 15 - Python ve Weka sonuçlarının karşılaştırılması**

Python DummyClassifier ve Weka ZeroR
sonuçları temel referans davranışı
açısından tutarlı bulunmuştur.

Bununla birlikte Python ve Weka deneyleri
tamamen eşdeğer değildir.

Python tarafında 5 katlı ve 2 tekrarlı
çapraz doğrulama, Weka tarafında ise
tek 10 katlı çapraz doğrulama kullanılmıştır.

Python'da Average Precision, Weka'da
ise PRC Area raporlanmıştır.

Bu ölçütlerin hesaplanması aynı
olmadığından sonuçlar doğrudan
sayısal eşitlik beklentisiyle
karşılaştırılmamıştır.

Ayrıca lojistik regresyon ve J48 farklı
öğrenme algoritmalarıdır.

Python deneylerinde eksik değer doldurma
ve kodlama Pipeline içerisinde yapılırken,
Weka J48 eksik öznitelik değerlerini
kendi mekanizmasıyla ele almıştır.

Bu nedenle Weka deneyleri, Python
modellerinin birebir yeniden üretimi
olarak değil, bağımsız bir yöntem
ve baseline kontrolü olarak yorumlanmıştır.

**Karar 16 – Deneylerin sınırlılıklarının belirtilmesi**

Çalışmanın başlıca sınırlılıkları şunlardır:

- Geliştirme kümesindeki katmanlı çapraz
  doğrulama kronolojik sırayı korumamaktadır.

  test bölümleri arasında belirgin biçimde
  farklılaşmaktadır.
- Day değişkeni, kodlama deneyinde
  kategorik kabul edilmiştir; 31 benzersiz
  değer yüksek kardinalitenin sınırlı
  bir örneğidir.
- `pdays=-1` özel kodu korunmuştur; bunun
  sayısal uzaklık olarak yorumlanmasının
  doğrusal modeller üzerindeki etkisi
  ayrıca araştırılmamıştır.
- Weka ve Python deneylerinde farklı
  sınıflandırıcılar, eksik değer işleme
  mekanizmaları ve değerlendirme
  protokolleri kullanılmıştır.
- Veri sızıntısı deneyleri, hatalı işlemlerin
  yöntemsel sonuçlarını göstermek amacıyla
  tasarlanmıştır; her sızıntı türünün
  mutlaka performansı artırdığı
  varsayılmamıştır.

Geliştirme kümesinde pozitif sınıf oranı
yaklaşık `%6,73` iken ayrılmış ileri dönem
test bölümünde yaklaşık `%31,58`'dir.

Bu fark, veri dağılımının bölümler arasında
değiştiğini göstermektedir.

Dolayısıyla geliştirme kümesindeki
çapraz doğrulama sonuçlarından hareketle
ileri dönem performansı hakkında kesin

### 4.2.1 Model karşılaştırması
**Dietterich (1998), okuma notu:** Tek bölmenin farkı bölme şansına duyarlıdır. Örtüşen CV katlarının skorlarına sıradan bağımsız t testi uygulamak yanlış belirsizlik verebilir. Burada ortak katlarla model karşılaştırması, ayrıca seçime katılmayan tek kronolojik test kümesinde eşleştirilmiş McNemar testi yapılır. Test sınıflandırma hatasını ölçer; AP sıralamasındaki farkın testi değildir.

Parametreler önceden sabittir: ağaç derinliği 6/min yaprak 30 yorumlanabilirlik ve varyans için; k=21 uzaklık gürültüsünü azaltmak için; orman 100 ağaç/min yaprak 10 maliyet-varyans dengesi için; histogram boosting 100 iterasyon/15 yaprak küçük etkileşimler için. Bunların optimal olduğu ileri sürülmez; tuning ayrı deneydir.



Bu bölüm Hakan’ın Bölüm 1 verisini ve ağırlıksız lojistiğini kullanır. Ağaç ve ormanda balanced ağırlığı korunur; algoritma ailesinin saf etkisi sınanmaz.


```python
modeller={
"Baseline":hat(DummyClassifier(strategy="prior")),
"Lojistik":hat(lojistik),
"Agac":hat(DecisionTreeClassifier(max_depth=6,min_samples_leaf=30,class_weight="balanced",random_state=TOHUM)),
"kNN":hat(KNeighborsClassifier(n_neighbors=21,n_jobs=1)),
"Orman":hat(RandomForestClassifier(n_estimators=100,min_samples_leaf=10,class_weight="balanced",random_state=TOHUM,n_jobs=1)),
"Artirma":hat(HistGradientBoostingClassifier(max_iter=100,max_leaf_nodes=15,random_state=TOHUM))}
for ad, model in modeller.items():
    print("Çalışıyor:",ad,flush=True)
    degerlendir(ad,model)
puanlar=pd.DataFrame(kayitlar)
model_puanlari=puanlar[puanlar.deney.isin(modeller)]
ozet=model_puanlari.groupby("deney")[list(olcutler)].agg(["mean","std"])
# mean ve standart sapma hesaplama
display(ozet)
ozet.to_csv(cikti/"model_ozeti.csv")
model_puanlari.boxplot(column="AP",by="deney",figsize=(8,4)); plt.ylabel("AP (0–1)"); plt.xlabel("Model"); plt.title("Aynı 5×2 katlarda AP dağılımı"); plt.suptitle(""); plt.tight_layout(); plt.show()
```

Model seçimi yukarıdaki güncel CV AP tablosuna dayanır. Ortalama ± örneklem standart sapması, örtüşen katların oynaklığıdır; güven aralığı veya anlamlılık testi değildir. Pozitif yakalama, kesinlik ve F1 ayrıca okunmalıdır. En yüksek AP, en iyi kalibrasyon veya 0,5 eşiğinde en yüksek F1 anlamına gelmez.


```python
olcek_kayitlari=[]
# model başarı etkilerini inceleme
for ad,karsilastirma_modeli,olcekler in [("Lojistik",lojistik,["standart","dayanikli"]),("kNN",KNeighborsClassifier(n_neighbors=21,n_jobs=1),["standart","dayanikli","yok"])]:
    for olcek in olcekler:
        if olcek=="standart":
            degerler=model_puanlari.loc[model_puanlari.deney.eq(ad),"AP"].to_numpy()
        else:
            degerler=cross_validate(hat(karsilastirma_modeli,olcekleme=olcek),egitim,sonuc,cv=katlar,scoring="average_precision",n_jobs=1,error_score="raise")["test_score"]
        for kat,deger in enumerate(degerler):olcek_kayitlari.append({"model":ad,"olcek":olcek,"kat":kat,"AP":deger})
olcek_puanlari=pd.DataFrame(olcek_kayitlari)
display(olcek_puanlari.groupby(["model","olcek"]).AP.agg(["mean","std"]))
olcek_puanlari.to_csv(cikti/"olcekleme_duyarliligi.csv",index=False)
```

## 4.2.2 Ağacın alan yorumu ve optimist yanlılık
İlk iki düzeydeki koşullar birlikte okunur. One-hot eşikleri 0.5 kategori yok/var anlamındadır; sayısal eşikler standartlaştırılmış birimdedir, ekonomik sınır diye yorumlanamaz. İç içe CV'de her dış eğitim kümesinde GridSearch yeniden kurulur; dış doğrulama tuning görmez.

**Kesinlik tanımı:** Hiç pozitif tahmin yoksa kesinlik TP/(TP+FP) tanımsızdır. sklearn skorlama varsayılanıyla 0 raporlar; WEKA  da `?` ıle gösterebilir. Bu araç sözleşmesidir; modelin pozitif kesinliğinin gerçekten ölçüldüğü anlamına gelmez.


```python
modeller["Agac"].fit(egitim,sonuc)
# karar agacı egitme
isimler=modeller["Agac"].named_steps["hazirla"].get_feature_names_out()
print(export_text(modeller["Agac"].named_steps["model"],feature_names=list(isimler),max_depth=2))
ic_katlar=StratifiedKFold(3,shuffle=True,random_state=TOHUM)
aramali=GridSearchCV(modeller["Agac"],{"model__max_depth":[3,6,9],"model__min_samples_leaf":[30,100]},cv=ic_katlar,scoring="average_precision",n_jobs=1)
aramali.fit(egitim,sonuc)
ic_iyi=aramali.best_score_
dis_skor=cross_validate(aramali,egitim,sonuc,cv=katlar,scoring="average_precision",n_jobs=1,error_score="raise")["test_score"]
print("GridSearch best_score_",ic_iyi,"İç içe dış AP",dis_skor.mean(),"±",dis_skor.std(ddof=1),"fark",ic_iyi-dis_skor.mean())
```

GridSearch iç skoru bağımsız test değildir. Dış katlar aramaya katılmaz; negatif iç–dış fark mümkündür. Gösterilen ağaçtaki sayısal eşikler ölçeklenmiş birimdedir; one-hot 0,5 eşikleri kategori varlığını ayırır. Bunlar nedensel veya ekonomik karar eşikleri değildir.


```python
sirali=model_puanlari.groupby("deney").AP.mean().sort_values(ascending=False)
ilk_iki=sirali.index[:2].tolist()
tahminler={}
ileri_kayitlari=[]
for ad in ilk_iki:
    modeller[ad].fit(egitim,sonuc)
    olasilik=modeller[ad].predict_proba(gelecek)[:,1]
    tahminler[ad]=(olasilik>=.5).astype(int)
    ileri_kayitlari.append({"model":ad,"AP":average_precision_score(gelecek_sonuc,olasilik),"F1":f1_score(gelecek_sonuc,tahminler[ad]),"hata_orani":np.mean(tahminler[ad]!=gelecek_sonuc.to_numpy()),"prevalans":gelecek_sonuc.mean()})
ileri_ozet=pd.DataFrame(ileri_kayitlari).set_index("model")
display(ileri_ozet)
ileri_ozet.to_csv(cikti/"ileri_donem_ozeti.csv")
yanlis_bir=tahminler[ilk_iki[0]]!=gelecek_sonuc.to_numpy()
yanlis_iki=tahminler[ilk_iki[1]]!=gelecek_sonuc.to_numpy()
b=int((yanlis_bir & ~yanlis_iki).sum()); c=int((~yanlis_bir & yanlis_iki).sum())
p_degeri=binomtest(b,b+c,.5).pvalue if b+c else 1.
print("McNemar exact",ilk_iki,"b",b,"c",c,"p",p_degeri)
from scipy.stats import chi2
ki_kare_duzeltilmis=max(abs(b-c)-1,0)**2/(b+c) if b+c else 0.
p_yaklasik=chi2.sf(ki_kare_duzeltilmis,df=1)
print("Ders 7 süreklilik düzeltmeli ki-kare",ki_kare_duzeltilmis,"p",p_yaklasik)
en_iyi_ad=ilk_iki[0]; en_iyi=modeller[en_iyi_ad]
joblib.dump(en_iyi,cikti/"p1_model.joblib")
veri.to_csv(cikti/"p1_anlamsal_temiz_veri.csv",index=False)
pd.DataFrame({"satir":veri.index,"bolum":np.where(veri.index<sinir,"gelistirme","ileri_donem")}).to_csv(cikti/"bolme.csv",index=False)
np.savez(cikti/"cv_katlari.npz",**{f"egitim_{i}":a for i,(a,b) in enumerate(katlar)},**{f"dogrulama_{i}":b for i,(a,b) in enumerate(katlar)})

print("b: birinci model yanlis, ikinci dogru; c: birinci dogru, ikinci yanlis")
print("Birinci eksi ikinci hata orani", (b-c)/len(gelecek_sonuc))
```

Modeller daha önce aynı veri üzerinde seçildiğinden p değerleri bağımsız doğrulama değildir.

Kaynak: Dietterich (1998), *Approximate Statistical Tests for Comparing Supervised Classification Learning Algorithms*. https://doi.org/10.1162/089976698300017197


```python
from sklearn.base import clone
from sklearn.model_selection import RepeatedStratifiedKFold
from scipy.stats import t as student_t

# 1) Aynı dış katlarla GridSearch: mevcut arama nesnesinin bir kopyası.
aramali_ayni = clone(aramali).set_params(cv=katlar)
aramali_ayni.fit(egitim, sonuc)
ic_iyi_ayni = float(aramali_ayni.best_score_)
print(f"Aynı katlarda GridSearch AP: {ic_iyi_ayni:.6f}")
print(f"İç içe dış AP: {dis_skor.mean():.6f}; fark: {ic_iyi_ayni-dis_skor.mean():+.6f}")

# 2) 5×2cv: iki algoritma, beş tekrar, her tekrarda iki yön.
bes_iki_katlar = RepeatedStratifiedKFold(n_splits=2, n_repeats=5, random_state=TOHUM)
bes_iki_kayitlari = []
for no, (tr, te) in enumerate(bes_iki_katlar.split(egitim, sonuc)):
    assert len(np.intersect1d(tr, te)) == 0
    X_tr, X_te = egitim.iloc[tr], egitim.iloc[te]
    y_tr, y_te = sonuc.iloc[tr], sonuc.iloc[te]
    ap, hata = [], []
    for model_adi in ilk_iki:
        model_52 = clone(modeller[model_adi])
        model_52.fit(X_tr, y_tr)
        olas_52 = model_52.predict_proba(X_te)[:, 1]
        ap.append(average_precision_score(y_te, olas_52))
        hata.append(np.mean((olas_52 >= 0.5).astype(int) != y_te.to_numpy()))
    bes_iki_kayitlari.append({"tekrar": no // 2 + 1, "yon": no % 2 + 1,
                             "AP": float(ap[0]-ap[1]), "hata_0.5": float(hata[0]-hata[1])})
    print(f"5×2cv: tekrar {no//2+1}/5, yön {no%2+1}/2", flush=True)
bes_iki_farklar = pd.DataFrame(bes_iki_kayitlari)
assert len(bes_iki_farklar) == 10

def dietterich_istatistik(farklar):
    p_ij = np.asarray(farklar, dtype=float).reshape(5, 2)
    s2 = ((p_ij - p_ij.mean(axis=1, keepdims=True))**2).sum(axis=1)
    payda = float(np.sqrt(s2.sum()/5))
    # Sıfır varyansı, geçerli test istatistiği gibi raporlamayın.
    if payda == 0:
        raise ValueError("5×2cv fark varyansı sıfır; Dietterich testi tanımsız.")
    t_degeri = float(p_ij[0,0] / payda)
    p_degeri_52 = float(2 * student_t.sf(abs(t_degeri), df=5))
    return {"ortalama_fark": float(p_ij.mean()), "t": t_degeri, "p": p_degeri_52}

bes_iki_tablo = pd.DataFrame({olcut: dietterich_istatistik(bes_iki_farklar[olcut])
                              for olcut in ("AP", "hata_0.5")}).T
assert bes_iki_tablo["p"].between(0,1).all()
display(bes_iki_farklar)
display(bes_iki_tablo)
bes_iki_farklar.to_csv(cikti/"dietterich_5x2cv_farklar.csv", index=False)
bes_iki_tablo.to_csv(cikti/"dietterich_5x2cv_ozet.csv")
```

### Veri küpünde %95 güven aralıkları
Oranlar için Wilson, kredi=yes eksi no farkı için Wilson sınırlarını birleştiren Newcombe yöntemi kullanılır. Bunlar bağımsız binom kayıt varsayımı altında noktasal aralıklardır; çoklu seçim, dönem kayması ve müşteri tekrarlarının belirsizliğini kapsamaz. Formüller scipy/numpy ile hesaplanır.

Normal yaklaşımın küçük hücrelerde ve sınır değerlere yakın oranlarda sorun yaratabilmesi nedeniyle Wilson aralığı tercih edilmiştir. İki grubun oran farkı için ayrı aralıkları gözle karşılaştırmak yerine, doğrudan farkı hedefleyen Newcombe aralığı kullanılır. Bu yaklaşım, yön ile tahmin hassasiyetini ayrı değerlendirir.

```python
# EK ANALIZ: Eski kup sonuclari degistirilmeden belirsizlik hesaplanir.
from scipy.stats import norm

def wilson_araligi(olumlu, kayit, alfa=0.05):
    olumlu=np.asarray(olumlu,dtype=float); kayit=np.asarray(kayit,dtype=float)
    if np.any(kayit<=0) or np.any(olumlu<0) or np.any(olumlu>kayit):
        raise ValueError("Gecerli olumlu sayisi ve pozitif kayit sayisi gerekli")
    z=norm.isf(alfa/2); oran=olumlu/kayit
    payda=1+z*z/kayit
    merkez=(oran+z*z/(2*kayit))/payda
    yari=z*np.sqrt(oran*(1-oran)/kayit+z*z/(4*kayit*kayit))/payda
    return np.clip(merkez-yari,0,1),np.clip(merkez+yari,0,1)

def oran_ozeti(tablo, boyutlar):
    ozet=tablo.groupby(boyutlar,observed=True,dropna=False).abonelik.agg(kayit="count",olumlu="sum",oran="mean")
    ozet["alt_95"],ozet["ust_95"]=wilson_araligi(ozet.olumlu,ozet.kayit)
    ozet["aralik_genisligi"]=ozet.ust_95-ozet.alt_95
    return ozet

def newcombe_farki(olumlu_bir,kayit_bir,olumlu_iki,kayit_iki):
    oran_bir=olumlu_bir/kayit_bir; oran_iki=olumlu_iki/kayit_iki
    alt_bir,ust_bir=wilson_araligi(olumlu_bir,kayit_bir)
    alt_iki,ust_iki=wilson_araligi(olumlu_iki,kayit_iki)
    fark=oran_bir-oran_iki
    return fark, fark-np.sqrt((oran_bir-alt_bir)**2+(ust_iki-oran_iki)**2), fark+np.sqrt((ust_bir-oran_bir)**2+(oran_iki-alt_iki)**2)

kup_genel_ga=oran_ozeti(kup,["konut_kredisi"])
kup_yas_ga=oran_ozeti(kup,["yas_grubu","konut_kredisi"])
kup_detay_ga=oran_ozeti(kup,["yas_grubu","konut_kredisi","egitim"])
print("Genel oranlar ve Wilson %95 araliklari (0-1)"); display(kup_genel_ga)
print("Yas katmanlari ve Wilson %95 araliklari (0-1)"); display(kup_yas_ga)
print("Uc boyutlu tum gozlenen hucreler (0-1)"); display(kup_detay_ga)
fark_satirlari=[]
for grup,tablo in [("tum_yaslar",kup)]+[(str(ad),alt_tablo) for ad,alt_tablo in kup.groupby("yas_grubu",observed=True)]:
    sayim=tablo.groupby("konut_kredisi").abonelik.agg(["sum","count"])
    fark,alt,ust=newcombe_farki(sayim.loc["yes","sum"],sayim.loc["yes","count"],sayim.loc["no","sum"],sayim.loc["no","count"])
    fark_satirlari.append({"yas_grubu":grup,"fark_yuzde_puan":100*fark,"alt_95_yuzde_puan":100*alt,"ust_95_yuzde_puan":100*ust,"sifiri_iceriyor":bool(alt<=0<=ust)})
kup_fark_ga=pd.DataFrame(fark_satirlari).set_index("yas_grubu")
print("Kredi yes eksi no: Newcombe %95 fark araliklari, yuzde puan"); display(kup_fark_ga)
for ad,satir in kup_fark_ga.iterrows():
    karar="sifiri icermiyor; bu varsayimlar altinda negatif farkla uyumlu" if satir.ust_95_yuzde_puan<0 else "sifiri iceriyor; farkin yonu bu aralikla kesinlestirilemiyor"
    display(Markdown(f"**Hesaplanan yorum — {ad}:** fark **{satir.fark_yuzde_puan:.2f} yüzde puan**, %95 aralık **[{satir.alt_95_yuzde_puan:.2f}; {satir.ust_95_yuzde_puan:.2f}]**; {karar}. Bu, kredi durumunun nedensel etkisi değildir."))
for ad,tablo in [("kup_genel_guven_araligi",kup_genel_ga),("kup_yas_guven_araligi",kup_yas_ga),("kup_detay_guven_araligi",kup_detay_ga),("kup_fark_guven_araligi",kup_fark_ga)]:
    tablo.to_csv(cikti/(ad+".csv"))
```

**Yorum ve karar:** Dört Newcombe aralığının hiçbiri sıfırı içermez: tüm yaşlarda −2,83 yp [−3,37; −2,30], 30 altında −8,50 [−10,84; −6,30], 30–59'da −1,96 [−2,51; −1,42], 60 ve üstünde −9,73 [−15,01; −3,40]. 60 ve üstü grubun aralığı en geniştir, çünkü burada yalnız 177 kredili kayıt vardır. **Yorum – roll-up doğruluğu, slice ve dice:** Sayımları birleştiren ağırlıklı roll-up genel oranı (%6,73) birebir verir. Bu, oranların toplanabilir bir ölçü olmadığını, fakat pay ve paydanın toplanabilir olduğunu gösterir. Slice tek bir boyutu sabitlerken dice birden fazla koşulla alt küp oluşturur. Dice'taki %25,0 değeri 31 olumlu kayda dayandığı için genel popülasyonun %25'inin abone olacağı biçiminde yorumlanmamalıdır. **Karar:** Segment karşılaştırmalarında oranla birlikte payda, olumlu sayısı ve mümkünse belirsizlik aralığı sunulmalıdır.

```python
# Drill-down ile Simpson kontrolünün genişletilmesi: aynı karşılaştırma eğitim katmanlarında ve yaş × eğitim hücrelerinde.
genel_fark=genel["yes"]-genel["no"]
egitim_katman=kup.groupby(["egitim","konut_kredisi"],observed=True,dropna=False).abonelik.agg(["count","mean"]).unstack()
egitim_katman["yes_eksi_no"]=egitim_katman[("mean","yes")]-egitim_katman[("mean","no")]
print("Eğitim katmanları (oran 0–1, fark = kredi yes − no)"); display(egitim_katman)
uc_boyut=kup.groupby(["yas_grubu","egitim","konut_kredisi"],observed=True,dropna=False).abonelik.agg(["count","mean"]).unstack()
uc_boyut["yes_eksi_no"]=uc_boyut[("mean","yes")]-uc_boyut[("mean","no")]
uc_boyut["en_kucuk_hucre"]=uc_boyut[[("count","no"),("count","yes")]].min(axis=1)
print("Yaş × eğitim hücreleri"); display(uc_boyut)
ters_egitim=int((np.sign(egitim_katman.yes_eksi_no)==-np.sign(genel_fark)).sum())
ters_uc=uc_boyut.loc[np.sign(uc_boyut.yes_eksi_no)==-np.sign(genel_fark)]
display(Markdown(f"**Genişletilmiş Simpson kontrolü:** genel fark {100*genel_fark:.2f} yüzde puan. Eğitim katmanlarında yönü ters dönen katman: **{ters_egitim}/{len(egitim_katman)}**. "
    f"Yaş × eğitim hücrelerinde ters dönen: **{len(ters_uc)}/{len(uc_boyut)}**"
    + (f" ({'; '.join(f'{i[0]}/{i[1]}: {100*r.yes_eksi_no.iloc[0]:+.2f} yp, en küçük hücre {int(r.en_kucuk_hucre.iloc[0])}' for i,r in ters_uc.groupby(level=[0,1],observed=True))})" if len(ters_uc) else "")
    + ". Tüm katmanlarda tersine dönme olmadığı için Simpson paradoksu görülmez."))
egitim_katman.to_csv(cikti/"simpson_egitim_katmanlari.csv"); uc_boyut.to_csv(cikti/"simpson_yas_egitim_hucreleri.csv")
```

## 4.3.2 Birliktelik · Seçenek B

Yönerge 4.3 uyarınca Seçenek B seçilmiştir: veri alışveriş sepeti ya da oturum içermediği için her pazarlama kaydı ayrı işlem, her öznitelik-değer çifti ayrı öğe olarak tanımlanır. Böylece örneğin `ay=mar` ve `konut_kredisi=no` aynı kayıtta birlikte yer alabilir. Hedef, yalnız kuralın sonuç tarafında `hedef=yes` olarak seçilir. Görüşme tamamlandıktan sonra bilinen süre analize katılmamıştır.

**Ayrıklaştırma gerekçesi:** Sürekli değerlerin tek tek öğe yapılması desteklerin parçalanmasına ve çok sayıda nadir kategoriye yol açabilir. Bu nedenle geliştirme verisinde üçte birlik kantiller kullanılır. Yaş sınırları 35 ve 45, bakiye sınırları 148 ve 880'dir. Kampanya sayısındaki yinelenen kantiller nedeniyle iki dilim (≤3 ve >3), önceki temas sayısındaki sıfır yığılması nedeniyle de iki dilim (0 ve pozitif) oluşmuştur. Küpteki 30 ve 60 yaş sınırlarının amacı alan temelli karşılaştırma yapmaktır; birliktelikteki farklı sınırlar ise öğelere yeterli kayıt desteği sağlamaya yöneliktir. Sonsuz uçlar gelecekteki değerleri de kapsamaya elverişlidir, ancak ileri dönem sınırları öğrenmek için kullanılmaz. Eksik kategorik değerler `eksik` olarak temsil edilir.

**Kural seçim protokolü:** Apriori ile en çok üç öğeli kümeler üretilir (hedef ve en fazla iki öncül); kısa kuralların yorumlanabilirliği ve arama maliyeti gözetilir. İlk keşifte 0,01 destek ve 0,20 güven ile yeterli pozitif kural elde edilemediği için ay öğesi eklenmiş ve destek 0,002'ye indirilmiştir. Bu karar veri gözlendikten sonra verildiğinden keşifseldir. Destek 0,002, 0,005 ve 0,01 için örüntü sayısı/süre duyarlılığı incelenir; bu çalışma Seçenek A'daki FP-Growth veya yapay null-işlem deneyi gerekliliğini üstlendiğini iddia etmez.

**Wu, Chen ve Han (2010) ile yöntem bağlantısı:** Çalışmanın önerdiği ilgi ölçütleri çerçevesinde tek bir sıralama yerine güven, lift, Kulczynski ve IR birlikte yorumlanır. Destek sıklığı, güven koşullu abonelik oranını, lift taban orana göre zenginleşmeyi, Kulczynski iki yönlü koşullu oranların dengesini ve IR öncül-sonuç yaygınlık farkını açıklar. Null-invariance özelliği aşağıda kavramsal olarak ele alınır; bu defterde yapay kayıt eklenerek ayrıca deney yapılmamıştır.

```python
from mlxtend.frequent_patterns import apriori, association_rules
import warnings
# mlxtend tum DeprecationWarning bildirimlerini acar; yalniz Colab istemci uyarisini sustur.
warnings.filterwarnings("ignore", message=r"datetime\.datetime\.utcnow\(\) is deprecated", category=DeprecationWarning)
islem=egitim[["meslek","egitim","konut_kredisi","bireysel_kredi","iletisim","onceki_sonuc","ay"]].fillna("eksik").copy()
sinirlar={}
for sutun in ["yas","bakiye","kampanya","onceki_sayi"]:
    _, kesimler=pd.qcut(egitim[sutun],q=3,retbins=True,duplicates="drop")
    kesimler[0]=-np.inf; kesimler[-1]=np.inf
    if len(kesimler)==2 and sutun=="onceki_sayi": kesimler=np.array([-np.inf,0,np.inf])
    sinirlar[sutun]=kesimler.tolist()
    islem[sutun]=pd.cut(egitim[sutun],kesimler,include_lowest=True,labels=["dilim_"+str(i+1) for i in range(len(kesimler)-1)]).astype(str)
islem["hedef"]=sonuc.map({0:"no",1:"yes"})
sepet=pd.get_dummies(islem.astype(str),prefix_sep="=",dtype=bool)
zamanlar=[]
for destek in [.002,.005,.01]:
    basla=time.perf_counter(); kumeler=apriori(sepet,min_support=destek,use_colnames=True,max_len=3)
    zamanlar.append({"destek":destek,"saniye":time.perf_counter()-basla,"oruntu":len(kumeler)})
    if destek==.002: temel_kumeler=kumeler
kurallar=association_rules(temel_kumeler,metric="confidence",min_threshold=.20)
kurallar=kurallar[kurallar.consequents.apply(lambda deger:deger==frozenset(["hedef=yes"]))].copy()
kurallar["kulczynski"] = .5*(kurallar["confidence"]+kurallar["support"]/kurallar["consequent support"])
kurallar["dengesizlik"] = abs(kurallar["antecedent support"]-kurallar["consequent support"])/(kurallar["antecedent support"]+kurallar["consequent support"]-kurallar["support"])
display(pd.DataFrame(zamanlar)); print("Kantil sınırları",sinirlar)
for olcut in ["confidence","lift","kulczynski"]:
    print("İlk 10:",olcut); display(kurallar.sort_values(olcut,ascending=False)[["antecedents","support","confidence","lift","kulczynski","dengesizlik"]].head(10))
kurallar.to_csv(cikti/"hedef_kurallari.csv",index=False)
nadir_ogeler=set(sepet.columns[sepet.mean()<.05])
nadir=kurallar.loc[kurallar.antecedents.apply(lambda deger:bool(set(deger)&nadir_ogeler)).astype(bool)]
print("%5 altı öğe içeren kurallar"); display(nadir.sort_values("lift",ascending=False).head(10))
```

```python
# Kural kümesinin yapısal kontrolü: dilim sayıları ve aynı kayıt kümesini kapsayan kurallar.
print("Dilim sayıları:",{s:len(k)-1 for s,k in sinirlar.items()})
esdeger=bool((islem.onceki_sonuc.eq("not_applicable")==islem.onceki_sayi.eq("dilim_1")).all())
icerme=bool(islem.loc[islem.onceki_sonuc.eq("success"),"onceki_sayi"].eq("dilim_2").all())
print("onceki_sonuc=not_applicable ⇔ onceki_sayi=dilim_1:",esdeger,"| onceki_sonuc=success ⊂ onceki_sayi=dilim_2:",icerme)
def kapsam_imzasi(oncul):
    maske=np.ones(len(islem),dtype=bool)
    for oge in oncul:
        sutun,deger=oge.split("=",1); maske&=islem[sutun].astype(str).eq(deger).to_numpy()
    return hashlib.sha1(np.packbits(maske).tobytes()).hexdigest()
imzalar=kurallar.antecedents.apply(kapsam_imzasi)
tekrar_gruplari=kurallar.groupby(imzalar).antecedents.apply(lambda s:[" & ".join(sorted(x)) for x in s])
tekrar_gruplari=tekrar_gruplari[tekrar_gruplari.str.len()>1]
farkli_kural=imzalar.nunique()
print(f"Kural sayısı {len(kurallar)}; farklı kayıt kümesine karşılık gelen kural {farkli_kural}. Aynı kayıtları kapsayan gruplar:")
for grup in tekrar_gruplari: print("  ", " ≡ ".join(grup))
```

### Lift ile Kulczynski farkı: nadir ve yaygın öncüller

Pozitif hedefli kısa kuralların tamamı yüksek güven filtresinden geçtiği için benzer seçici profilleri içermektedir. İlgi ölçütleri arasındaki farkı daha geniş bir örnek uzayında incelemek amacıyla tüm tek öncüllü kurallar, güven filtresi olmadan fakat aynı asgari ortak destekle (en az 73 kayıt) hesaplanır. Bu, ana 12 kuralı değiştiren ikinci bir seçim değildir; ölçüt duyarlılığı analizidir.

`A` öncül, `B` abonelik=yes, `n_A` ve `n_B` marjinal sayımlar, `n_AB` birlikte görülme sayısı ve `N` toplam kayıt sayısı olsun:

| Ölçüt | Hesap | Yorum |
|---|---|---|
| Destek | n_AB / N | Ortak olayın veri içindeki sıklığı |
| Güven | n_AB / n_A | Öncülü sağlayanlarda abonelik oranı |
| Lift | N × n_AB / (n_A × n_B) | Bağımsızlığa göre göreli zenginleşme; 1 referanstır |
| Kulczynski | ½ × (n_AB/n_A + n_AB/n_B) | Güven ile abonelerin kapsanmasının simetrik ortalaması |
| IR | abs(n_A − n_B) / (n_A + n_B − n_AB) | İki olayın yaygınlığı arasındaki dengesizlik |

Wu, Chen ve Han (2010) bağlantısı: Her iki olayı da içermeyen kayıtlar eklenirse n_A, n_B ve n_AB sabit kalır, N artar. Bu nedenle lift değişir; Kulczynski ve IR değişmez (null-invariance). Güven de bu özel işlem altında sabittir, fakat yönlüdür. Bu özellik farklı veri büyüklüklerinde ilgisiz kayıtların ölçüte etkisini anlamaya yarar; burada cebirsel yorum olarak kullanılmış, ayrı bir null-işlem deneyi yapılmamıştır. IR'nin 0'a yakınlığı iki marjinal sayımın benzerliğini, 1'e yakınlığı belirgin dengesizliği gösterir; tek başına pozitif ilişkinin testi değildir.

```python
taban_oran=sonuc.mean(); kayit_sayisi=len(islem); evet=islem.hedef.eq("yes").to_numpy(); min_ortak=int(np.ceil(.002*kayit_sayisi))
tek_satirlar=[]
for sutun in islem.columns.drop("hedef"):
    for deger in sorted(islem[sutun].astype(str).unique()):
        maske=islem[sutun].astype(str).eq(deger).to_numpy(); a=int(maske.sum()); ab=int((maske&evet).sum())
        if ab<min_ortak: continue
        guven=ab/a; kapsama=ab/evet.sum()
        tek_satirlar.append({"oncul":f"{sutun}={deger}","oncul_destegi":a/kayit_sayisi,"guven_P(evet|A)":guven,"kapsama_P(A|evet)":kapsama,
            "lift":guven/taban_oran,"kulczynski":.5*(guven+kapsama),"IR":abs(a/kayit_sayisi-taban_oran)/(a/kayit_sayisi+taban_oran-ab/kayit_sayisi),"nadir_oge(<%5)":a/kayit_sayisi<.05})
tek_oncul=pd.DataFrame(tek_satirlar)
tek_oncul["lift_sirasi"]=tek_oncul.lift.rank(ascending=False,method="min").astype(int)
tek_oncul["kulczynski_sirasi"]=tek_oncul.kulczynski.rank(ascending=False,method="min").astype(int)
print("Lift'e göre ilk 8"); display(tek_oncul.sort_values("lift",ascending=False).head(8).round(4))
print("Kulczynski'ye göre ilk 8"); display(tek_oncul.sort_values("kulczynski",ascending=False).head(8).round(4))
spearman_lk=tek_oncul[["lift","kulczynski"]].corr(method="spearman").iloc[0,1]
nadir_ust=tek_oncul.sort_values("lift",ascending=False).iloc[0]; kulc_ust=tek_oncul.sort_values("kulczynski",ascending=False).iloc[0]
display(Markdown(f"**Hesaplanan karşılaştırma:** {len(tek_oncul)} tek öncüllü kuralda lift ile Kulczynski sıralamalarının Spearman korelasyonu **{spearman_lk:.2f}**. "
    f"Lift'te birinci `{nadir_ust.oncul}` (destek %{100*nadir_ust.oncul_destegi:.2f}, lift {nadir_ust.lift:.2f}) Kulczynski'de **{int(nadir_ust.kulczynski_sirasi)}.** sıradadır, çünkü aboneliklerin yalnız %{100*nadir_ust['kapsama_P(A|evet)']:.1f}'ini kapsar. "
    f"Kulczynski'de birinci `{kulc_ust.oncul}` (destek %{100*kulc_ust.oncul_destegi:.1f}) ise lift {kulc_ust.lift:.2f} ile lift'te **{int(kulc_ust.lift_sirasi)}.** sıradadır; IR={kulc_ust.IR:.2f}."))
tek_oncul.to_csv(cikti/"lift_kulczynski_karsilastirma.csv",index=False)
fig,eksen=plt.subplots(figsize=(7,4.5))
for nadir_mi,renk,etiket in [(True,"tab:red","nadir öğe (<%5 destek)"),(False,"tab:blue","yaygın öğe")]:
    alt=tek_oncul[tek_oncul["nadir_oge(<%5)"]==nadir_mi]; eksen.scatter(alt.lift,alt.kulczynski,c=renk,label=etiket,alpha=.75)
for _,r in pd.concat([tek_oncul.nlargest(3,"lift"),tek_oncul.nlargest(3,"kulczynski")]).drop_duplicates("oncul").iterrows():
    eksen.annotate(r.oncul,(r.lift,r.kulczynski),fontsize=8,xytext=(4,3),textcoords="offset points")
eksen.axvline(1,color="gray",ls="--",lw=1); eksen.set_xlabel("Lift (birimsiz; 1 = bağımsızlık)"); eksen.set_ylabel("Kulczynski (0–1)")
eksen.set_title("Tek öncüllü öğe → abonelik kuralları: lift ve Kulczynski"); eksen.legend(); plt.tight_layout(); plt.show()
```

### Ek analiz · Birliktelik kurallarının güven değerine %95 aralık

Kural güveni P(abonelik=yes | öncül), öncülü sağlayan kayıtlar içindeki abonelik oranıdır. Her mevcut kural için öncül ve ortak sayıları yeniden sayılır, eski destek/güven ile eşitliği denetlenir ve güven için **Wilson %95 aralığı** hesaplanır. Payda tüm veri değil, öncülü sağlayan kayıtların sayısıdır. Önceki destek/güven/lift/Kulczynski değerleri ve kural seçimi korunur; aralık eklemek yeni kurallar keşfetmez ve mevcut sıralamayı yeniden optimize etmez.


```python
# EK ANALIZ: Her mevcut kuralin guveni icin Wilson araligi.
# Hicbir destek, guven esigi, kural sirasi veya eski sonuc degistirilmez.
kural_ga_satirlari=[]
for _,satir in kurallar.iterrows():
    maske=np.ones(len(islem),dtype=bool)
    for oge in satir.antecedents:
        sutun,deger=oge.split("=",1)
        maske &= islem[sutun].astype(str).eq(deger).to_numpy()
    oncul_sayisi=int(maske.sum())
    ortak_sayisi=int((maske & islem.hedef.eq("yes").to_numpy()).sum())
    guven=ortak_sayisi/oncul_sayisi
    assert np.isclose(guven,satir["confidence"])
    assert np.isclose(ortak_sayisi/len(islem),satir["support"])
    alt,ust=wilson_araligi(ortak_sayisi,oncul_sayisi)
    kural_ga_satirlari.append({"kural":" & ".join(sorted(satir.antecedents))+" -> hedef=yes", "oncul_sayisi":oncul_sayisi,"ortak_sayisi":ortak_sayisi,"destek":satir["support"],"guven":guven,"guven_alt_95":float(alt),"guven_ust_95":float(ust),"lift":satir["lift"],"kulczynski":satir["kulczynski"]})
kural_guven_ga=pd.DataFrame(kural_ga_satirlari).sort_values("kulczynski",ascending=False)
print("Mevcut pozitif kurallarin tamami; guven ve araliklar 0-1 olceginde")
display(kural_guven_ga.reset_index(drop=True))
for _,satir in kural_guven_ga.head(3).iterrows():
    display(Markdown(f"**Hesaplanan yorum — {satir.kural}:** {satir.ortak_sayisi}/{satir.oncul_sayisi} kayıtta abonelik; güven **%{100*satir.guven:.2f}**, noktasal %95 Wilson aralığı **[%{100*satir.guven_alt_95:.2f}; %{100*satir.guven_ust_95:.2f}]**. Bu aralık, sabit bir profil ve bağımsız kayıtlar varsayımına dayanır; çok sayıda adaydan kural seçmenin iyimserliğini veya gelecek dönemdeki değişimi kapsamaz."))
kural_guven_ga.to_csv(cikti/"kural_guven_araliklari.csv",index=False)
```

## WEKA · ortak veriyle bağımsız kontrol
Bölüm 1’deki ZeroR ve J48 (C=0,25, M=2) kontrolü korunur. Bu bölüm aynı hazırlanmış verinin Türkçe sütun adlarıyla ZeroR ve J48 (M=2) çalıştırmasını yapar; sonuçları güncel loglardan okuyun. C4.5/CART ve Python/WEKA katları eşdeğer değildir.

**Doğrulama yaklaşımı:** WEKA kontrolü, aynı hazırlanmış geliştirme verisi üzerinde iki farklı yazılım uygulamasının davranışını karşılaştırır. Sınıflandırmada çoğunluk tahmini ve ağaç davranışı referans alınırken birliktelikte birebir sayımlara odaklanılır. Yönerge 4.3'ün asıl bağımsız uygulama kontrolü Apriori'ye ilişkindir; bu nedenle J48 karşılaştırması tamamlayıcıdır, Apriori doğrulamasının yerine geçmez.

```python
def arff_yaz(tablo,yol):
    import csv
    def tirnak(deger):
        return "'"+str(deger).replace("\\","\\\\").replace("'","\\'")+"'"
    satirlar=["@relation banka"]
    for sutun in tablo:
        if pd.api.types.is_numeric_dtype(tablo[sutun]): tur="numeric"
        else: tur="{"+",".join(tirnak(deger) for deger in sorted(tablo[sutun].dropna().unique()))+"}"
        satirlar.append("@attribute "+tirnak(sutun)+" "+tur)
    satirlar.append("@data")
    for satir in tablo.itertuples(index=False,name=None):
        satirlar.append(",".join("?" if pd.isna(deger) else str(deger) if isinstance(deger,(int,float,np.number)) else tirnak(deger) for deger in satir))
    Path(yol).write_text("\n".join(satirlar),encoding="utf-8")
weka_veri=egitim.copy(); weka_veri["hedef"]=sonuc.map({0:"no",1:"yes"})
arff_yaz(weka_veri,cikti/"weka_siniflandirma.arff")
arff_yaz(islem,cikti/"weka_birliktelik.arff")
java_yolu=JAVA
weka_sinif_yolu=weka_classpath
for ad,secenek in [("ZeroR",["weka.classifiers.rules.ZeroR"]),("J48",["weka.classifiers.trees.J48","-C","0.25","-M","2"])]:
    komut=[java_yolu,"--add-opens","java.base/java.lang=ALL-UNNAMED","-cp",weka_sinif_yolu]+secenek+["-t",str(cikti/"weka_siniflandirma.arff"),"-x","10","-s","42"]
    islem_sonucu=subprocess.run(komut,capture_output=True,text=True)
    if islem_sonucu.returncode or "Exception" in islem_sonucu.stderr or "exception:" in islem_sonucu.stderr: raise RuntimeError(islem_sonucu.stderr)
    (cikti/("weka_"+ad+".txt")).write_text(islem_sonucu.stdout)
    print(ad,islem_sonucu.stdout[-4500:])
komut=[java_yolu,"--add-opens","java.base/java.lang=ALL-UNNAMED","-Xmx2g","-cp",weka_sinif_yolu,"weka.associations.Apriori","-t",str(cikti/"weka_birliktelik.arff"),"-A","-c","last","-T","0","-C","0.2","-M",str(np.ceil(.002*len(islem))/len(islem)),"-U","1","-N","2147483647"]
islem_sonucu=subprocess.run(komut,capture_output=True,text=True)
if islem_sonucu.returncode or "Exception" in islem_sonucu.stderr or "exception:" in islem_sonucu.stderr: raise RuntimeError(islem_sonucu.stderr)
(cikti/"weka_Apriori.txt").write_text(islem_sonucu.stdout)
print(islem_sonucu.stdout[:1500])
```
